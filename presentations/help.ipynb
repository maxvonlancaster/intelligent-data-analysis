```python
import re
import string
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
import nltk

# Завантаження ресурсів NLTK (виконується один раз)
nltk.download('stopwords')
nltk.download('punkt')

def clean_text(text):
        # Перевести в нижній регістр
        text = text.lower()

        # Видалити розділові знаки
        text = text.translate(str.maketrans('', '', string.punctuation))

        # Видалити числа
        text = re.sub(r'\d+', '', text)

        # Видалити спеціальні символи (залишаємо лише англійські літери та пробіли)
        text = re.sub(r'[^a-z\s]', '', text)

        # Видалити стоп-слова
        stop_words = set(stopwords.words('english'))
        text = ' '.join(word for word in word_tokenize(text) if word not in stop_words)

        return text

sample_text = "Hello! This is an example text with numbers 123 and punctuation!!!"
cleaned_text = clean_text(sample_text)
print(cleaned_text)  # очікуваний вивід: hello example text numbers
```

```python
from nltk.tokenize import word_tokenize, sent_tokenize

def tokenize_text(text):
        words = word_tokenize(text)
        sentences = sent_tokenize(text)
        return words, sentences

sample_text = "Natural Language Processing (NLP) is exciting! It enables computers to understand human language."
words, sentences = tokenize_text(sample_text)
print('Words:', words)
print('Sentences:', sentences)
```

```python
from nltk.stem import PorterStemmer, WordNetLemmatizer
from nltk.tokenize import word_tokenize
import nltk

nltk.download('wordnet')

def apply_stemming(text):
        stemmer = PorterStemmer()
        words = word_tokenize(text)
        stemmed_words = [stemmer.stem(word) for word in words]
        return stemmed_words

def apply_lemmatization(text):
        lemmatizer = WordNetLemmatizer()
        words = word_tokenize(text)
        lemmatized_words = [lemmatizer.lemmatize(word) for word in words]
        return lemmatized_words

sample_text = "The children are playing with their toys."
print('Stemmed:', apply_stemming(sample_text))
print('Lemmatized:', apply_lemmatization(sample_text))
```

```python
from nltk.tokenize import word_tokenize
from autocorrect import Speller
import nltk

nltk.download('punkt')

def normalize_text(text):
        spell = Speller(lang='en')
        normalization_dict = {
                "u": "you",
                "ur": "your",
                "r": "are",
                "pls": "please",
                "thx": "thanks"
        }
        words = word_tokenize(text.lower())
        normalized_words = [normalization_dict.get(word, spell(word)) for word in words]
        normalized_text = ' '.join(normalized_words)
        normalized_text = normalized_text.replace('!', '').replace('.', '').strip()
        return normalized_text

sample_text = "U r the best! Pls help me with this assignment. Thx!"
print(normalize_text(sample_text))
```

```python
import numpy as np

documents = [
        "In Inception, the plot unfolds with thrilling action, keeping audiences on edge through a series of intense and unexpected twists.",
        "The story of Inception keeps audiences hooked with its thrilling action and surprising twists, making every moment unforgettable.",
        "The Matrix transformation preserves the vector's direction, illustrating how linear combinations affect the overall system."
]

document_words = [doc.lower().split() for doc in documents]
vocab = sorted(set(sum(document_words, [])))
vocab_dict = {k: i for i, k in enumerate(vocab)}

X_tf = np.zeros((len(documents), len(vocab)), dtype=int)
for i, doc in enumerate(document_words):
        for word in doc:
                X_tf[i, vocab_dict[word]] += 1


idf = np.log(X_tf.shape[0] / X_tf.astype(bool).sum(axis=0))
X_tfidf = X_tf * idf
X_tfidf_norm = X_tfidf / np.linalg.norm(X_tfidf, axis=1)[:, None]
M = X_tfidf_norm @ X_tfidf_norm.T
print('Cosine similarity matrix:\n', M)
```

```python
import numpy as np
from sentence_transformers import SentenceTransformer, util

# 1. Load a pre-trained Transformer model (all-MiniLM-L6-v2 is fast and accurate)
model = SentenceTransformer('all-MiniLM-L6-v2')

# 2. Extract a vector for 'computer'
# sentence-transformers always takes a list of strings
vector = model.encode(['computer'])[0]
print('vector sample:', vector[:10])

# 3. Calculate similarity between 'woman' and 'man'
embeddings = model.encode(['woman', 'man'])
# util.cos_sim returns a matrix; we extract the scalar value
similarity = util.cos_sim(embeddings[0], embeddings[1]).item()
print('similarity woman/man:', similarity)

# 4. Set up the word analogy: King - Man + Woman
words_to_encode = ['king', 'man', 'woman']
word_vectors = model.encode(words_to_encode)

king = word_vectors[0]
man = word_vectors[1]
woman = word_vectors[2]

result_vector = king - man + woman

# 5. Find the most similar words to the result vector
# Note: Unlike GloVe, Transformer spaces are vastly large. We define a search vocabulary.
vocabulary = ['queen', 'princess', 'monarch', 'empress', 'man', 'computer', 'throne', 'kyiv']
vocab_embeddings = model.encode(vocabulary)

# Calculate cosine similarities between our target vector and the vocabulary
similarities = util.cos_sim(result_vector, vocab_embeddings)[0].numpy()

# Get the top 5 indices sorted by highest similarity
top_indices = np.argsort(similarities)[::-1][:5]

print('\nAnalogy results:')
for idx in top_indices:
    print(f"{vocabulary[idx]}: {similarities[idx]:.4f}")
```

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import ConfusionMatrixDisplay, accuracy_score

# Завантажуємо датасет
data = load_breast_cancer()
X = data.data[:, [3, 25]]  
y = data.target

# Розподіл на тренувальні та тестові дані
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42
)

# Скейлимо ознаки
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

# Тренуємо Random Forest
clf = RandomForestClassifier(
    n_estimators=100, max_depth=5, random_state=42
)
clf.fit(X_train_scaled, y_train)

# Оцінка моделі
y_pred = clf.predict(X_test_scaled)
print("Accuracy:", accuracy_score(y_test, y_pred))

# Візуалізація
x_min, x_max = X_train_scaled[:, 0].min() - 1, X_train_scaled[:, 0].max() + 1
y_min, y_max = X_train_scaled[:, 1].min() - 1, X_train_scaled[:, 1].max() + 1
xx, yy = np.meshgrid(
    np.linspace(x_min, x_max, 300),
    np.linspace(y_min, y_max, 300)
)

Z = clf.predict(np.c_[xx.ravel(), yy.ravel()])
Z = Z.reshape(xx.shape)

plt.figure(figsize=(8, 6))
plt.contourf(xx, yy, Z, alpha=0.3, cmap='coolwarm')
plt.scatter(X_test_scaled[:, 0], X_test_scaled[:, 1],
            c=y_test, cmap='coolwarm', edgecolors='k', s=50)
plt.xlabel(data.feature_names[3])
plt.ylabel(data.feature_names[25])
plt.title("Random Forest Classifier Decision Boundary")
plt.show()

# Матриця похибок
ConfusionMatrixDisplay.from_predictions(y_test, y_pred, cmap='Blues')
plt.title("Random Forest Confusion Matrix")
```

```python
import numpy as np
import matplotlib.pyplot as plt

# 1. Задаємо дані
x = np.linspace(0.1, 2.9, 100)
bias_sq = 0.5 * ((3 - x) ** 3) + 0.5               # Зменшується зі складністю
variance = 0.5 * (x ** 3) + 0.5           # Зростає зі складністю
total_error = bias_sq + variance + 0.3  # Загальна помилка

# 2. Налаштування стилю
fig, ax = plt.subplots(figsize=(8, 5))

# Побудова ліній
ax.plot(x, total_error, color='black', linewidth=1.5, label='Загальна похибка')
ax.plot(x, variance, color='#2b939b', linewidth=1.5, label='Дисперсія')
ax.plot(x, bias_sq, color='#8b1a0e', linewidth=1.5, label=r'$\mathrm{Зсув}^2$')

# Пунктирна лінія оптимуму
optimum_x = x[np.argmin(total_error)]
ax.axvline(x=optimum_x, color='black', linestyle=':', linewidth=1.5)

# Налаштування осей (стрілки замість стандартних рамок)
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
ax.spines['left'].set_position(('data', 0))
ax.spines['bottom'].set_position(('data', 0))

# Підписи
ax.set_xlabel('Складність моделі', fontsize=12, labelpad=10)
ax.set_ylabel('Похибка', fontsize=12, labelpad=10)
ax.set_xticks([])
ax.set_yticks([])
ax.legend(frameon=False, fontsize=11)

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# 1. Задаємо 2D-координати для слів (імітація PCA/t-SNE проекції)
words = {
    'Man': np.array([1.0, 1.5]),
    'Woman': np.array([1.0, 4.0]),
    'King': np.array([4.5, 2.0]),
    'Queen': np.array([4.5, 4.5])
}

# Розрахунок результату векторної арифметики: King - Man + Woman
calculated_queen = words['King'] - words['Man'] + words['Woman']

# 2. Налаштування фігури
fig, ax = plt.subplots(figsize=(9, 6))

# Початкові вектори від центру (0,0)
for word, coord in words.items():
    ax.annotate('', xy=coord, xytext=(0, 0),
                arrowprops=dict(arrowstyle="-|>", color='#2b939b', lw=1.8, mutation_scale=15))
    # Підписи точок
    ax.text(coord[0] + 0.1, coord[1] + 0.1, word, fontsize=13, fontweight='bold')

# Векторні паралелограми (зсув гендеру: від Man до Woman та від King до Queen)
ax.annotate('', xy=words['Woman'], xytext=words['Man'],
            arrowprops=dict(arrowstyle="->", color='#8b1a0e', lw=1.5, ls='--'))

ax.annotate('', xy=words['Queen'], xytext=words['King'],
            arrowprops=dict(arrowstyle="->", color='#8b1a0e', lw=1.5, ls='--'))

# Текстовий підпис векторного зсуву
ax.text(0.3, 2.75, '+ (Woman - Man)\n[Гендерний вектор]', color='#8b1a0e', fontsize=10, fontstyle='italic')
ax.text(3.8, 3.25, '+ (Woman - Man)', color='#8b1a0e', fontsize=10, fontstyle='italic')

# Позначка для обчисленого результату (King - Man + Woman)
ax.scatter(calculated_queen[0], calculated_queen[1], color='black', s=50, zorder=5)

# 3. Оформлення осей та стилю
ax.set_xlim(-0.5, 6.0)
ax.set_ylim(-0.5, 5.5)

# Стилізація під науковий графік (вісь x та y зі стрілками)
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
ax.spines['left'].set_position(('data', 0))
ax.spines['bottom'].set_position(('data', 0))

# Підписи осей (Виміри векторного простору)
ax.set_xlabel('Семнтичний вимір 1 (напр. Статус / Роялті)', fontsize=11, labelpad=10)
ax.set_ylabel('Семнтичний вимір 2 (напр. Гендер)', fontsize=11, labelpad=10)

ax.set_xticks([])
ax.set_yticks([])

ax.set_title(r'$\mathrm{King} - \mathrm{Man} + \mathrm{Woman} \approx \mathrm{Queen}$', fontsize=14, pad=20)

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.decomposition import PCA

# Set seed
np.random.seed(42)

# Generate correlated 2D synthetic data
mean = [5, 5]
cov = [[1.5, 0.65], [0.65, 0.6]]
data = np.random.multivariate_normal(mean, cov, 1200)

# Perform PCA
pca = PCA(n_components=2)
pca.fit(data)

# Create figure
fig, ax = plt.subplots(figsize=(10, 6.5))

# Plot grid background effect
ax.grid(True, which='both', color='#e0e0e0', linestyle='-', linewidth=0.8)
ax.set_axisbelow(True)

# Scatter plot
ax.scatter(data[:, 0], data[:, 1], color='#f6c25e', alpha=0.75, s=18, edgecolors='none')

# Mean point
center = pca.mean_

# Principal Component axes
v1 = pca.components_[0] * np.sqrt(pca.explained_variance_[0])
v2 = pca.components_[1] * np.sqrt(pca.explained_variance_[1])

# Extended dashed red lines (Z and W axes)
scale_line = 3.5
ax.plot([center[0] - scale_line * v1[0], center[0] + scale_line * v1[0]],
        [center[1] - scale_line * v1[1], center[1] + scale_line * v1[1]],
        color='#d9534f', linestyle='--', linewidth=1.8, zorder=3)

ax.plot([center[0] - scale_line * v2[0], center[0] + scale_line * v2[0]],
        [center[1] - scale_line * v2[1], center[1] + scale_line * v2[1]],
        color='#d9534f', linestyle='--', linewidth=1.8, zorder=3)

# PC Vectors (Purple arrows)
ax.annotate('', xy=center + v1*1.2, xytext=center,
            arrowprops=dict(arrowstyle="-|>", color='#4a4398', lw=2.5, mutation_scale=18))
ax.annotate('', xy=center + v2*1.2, xytext=center,
            arrowprops=dict(arrowstyle="-|>", color='#4a4398', lw=2.5, mutation_scale=18))

# Text annotations
ax.text(center[0] + scale_line * v1[0] - 0.2, center[1] + scale_line * v1[1] + 0.1,
        'Z', fontsize=18, fontweight='bold', color='#333333')
ax.text(center[0] - scale_line * v2[0] - 0.2, center[1] - scale_line * v2[1] + 0.2,
        'W', fontsize=18, fontweight='bold', color='#333333')

ax.text(center[0] + v1[0]*1.2 - 0.1, center[1] + v1[1]*1.2 - 0.3,
        r'$\mathbf{PC_1}$', fontsize=13, color='#333333', fontweight='bold')
ax.text(center[0] + v2[0]*1.2 - 0.45, center[1] + v2[1]*1.2 - 0.1,
        r'$\mathbf{PC_2}$', fontsize=13, color='#333333', fontweight='bold')

# Main Axes (Purple thick lines with arrows)
xmin, xmax = 0.5, 9.5
ymin, ymax = 1.0, 9.0

ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
ax.spines['left'].set_color('#4a4398')
ax.spines['left'].set_linewidth(2.5)
ax.spines['bottom'].set_color('#4a4398')
ax.spines['bottom'].set_linewidth(2.5)

ax.set_xlim(xmin, xmax)
ax.set_ylim(ymin, ymax)

# Arrow heads on X and Y axes
ax.annotate('', xy=(xmax, ymin), xytext=(xmin, ymin),
            arrowprops=dict(arrowstyle="-|>", color='#4a4398', lw=2.5, mutation_scale=20))
ax.annotate('', xy=(xmin, ymax), xytext=(xmin, ymin),
            arrowprops=dict(arrowstyle="-|>", color='#4a4398', lw=2.5, mutation_scale=20))

# Custom X and Y Axis Labels
ax.text(xmin - 0.4, (ymin + ymax) / 2, 'Y', fontsize=18, fontweight='bold', color='#333333', va='center')
ax.text((xmin + xmax) / 2, ymin - 0.5, 'X', fontsize=18, fontweight='bold', color='#333333', ha='center')

# Title
ax.set_title('Метод Головних Компонент\n(Principal Component Analysis, PCA)', fontsize=22, fontweight='bold', color='#333333', pad=15)

# Remove numeric ticks to match original diagram
ax.set_xticks([])
ax.set_yticks([])

plt.tight_layout()
plt.savefig('pca_recreation.png', dpi=300)
```
